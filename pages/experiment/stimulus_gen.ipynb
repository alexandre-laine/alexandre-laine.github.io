```python
import matplotlib.pyplot as plt
import MotionClouds as mc
import numpy as np
```

```python
for th in np.linspace(0, np.pi, 10):
    print(f"th{th:.3f}")
```

```python
N_X = 1024
N_Y = 1024
N_frame = 1

fx, fy, ft = mc.get_grids(N_X, N_Y, N_frame)

for th in np.linspace(0, np.pi, 20):

    envelope = mc.envelope_gabor(fx, fy, ft,
        V_X=1., V_Y=0., B_V=.1,
        sf_0=.1, B_sf=.2,
        theta=th, B_theta=np.pi/32, 
        alpha=1.
    )

    cloud = mc.random_cloud(envelope)
    cloud = mc.rectif(cloud)[:, :, 0]

    for i in range(2):

        fig = plt.figure(
            figsize=(5, 5),
            dpi=200
        )
        plt.imshow(
            cloud,
            cmap="gray"
        )
        plt.axis("off")
        plt.tight_layout()
        fig.savefig(
            fname=f"./stim/test{i}th{th:.3f}.png"
        )
        plt.close()
```
